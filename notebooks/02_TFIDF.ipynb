```python
import pandas as pd

urdu_df = pd.read_csv("final_urdu_dataset.csv")

print(urdu_df.head())
print(urdu_df.columns)
```

```python
X_urdu = urdu_tfidf.transform(
    urdu_df["comment"].fillna("")
)

print("Urdu TF-IDF Shape:", X_urdu.shape)
```

```python
urdu_tfidf.transform(urdu_df["comment"].fillna(""))
```

```python
y_urdu = urdu_df["sentiment"]
```

```python
from sklearn.model_selection import train_test_split

X_train_urdu, X_test_urdu, y_train_urdu, y_test_urdu = train_test_split(
    X_urdu,
    y_urdu,
    test_size=0.20,
    random_state=42,
    stratify=y_urdu
)

print("Training data:", X_train_urdu.shape)
print("Testing data:", X_test_urdu.shape)
```

```python
from sklearn.linear_model import LogisticRegression

urdu_model = LogisticRegression(
    max_iter=1000,
    random_state=42
)

urdu_model.fit(X_train_urdu, y_train_urdu)

print("Urdu Logistic Regression model trained successfully!")
```

```python
y_pred_urdu = urdu_model.predict(X_test_urdu)

print(y_pred_urdu[:10])
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix
)

accuracy = accuracy_score(y_test_urdu, y_pred_urdu)
precision = precision_score(y_test_urdu, y_pred_urdu, average="weighted")
recall = recall_score(y_test_urdu, y_pred_urdu, average="weighted")
f1 = f1_score(y_test_urdu, y_pred_urdu, average="weighted")

print("Accuracy :", accuracy)
print("Precision:", precision)
print("Recall   :", recall)
print("F1 Score :", f1)

print("\nClassification Report:")
print(classification_report(y_test_urdu, y_pred_urdu))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test_urdu, y_pred_urdu))
```

```python
import joblib

joblib.dump(urdu_model, "urdu_logistic_regression.pkl")

print("Urdu model saved successfully!")
```
