```python
def normalize_roman_urdu(text):
    words = text.split()
    new_words = []

    for w in words:
        w_lower = w.lower()

        if w_lower in roman_normalization:
            new_words.append(roman_normalization[w_lower])
        else:

            w_clean = re.sub(r'(.)\1{2,}', r'\1', w_lower)

            new_words.append(w_clean)

    return " ".join(new_words)
```

```python
import pandas as pd
import random

# Step 1: Load your final dataset
df = pd.read_csv('combined_cleaned.csv')

# Step 2: Define basic Roman Urdu mapping for common words
roman_dict = {
    'یہ': 'ye',
    'ویڈیو': 'video',
    'بہت': 'bohat',
    'اچھی': 'achhi',
    'پسند': 'pasand',
    'نہیں': 'nahi',
    'کام': 'kaam',
    'زبردست': 'zabardast',
    'ٹھیک': 'theek',
    'کامیاب': 'kamyab',
    'ٹیم': 'team',
    'میچ': 'match',
    'ہاں': 'haan',
    'نہیں': 'nahi',
    'ہے': 'hai'
}

# Step 3: Function to convert Urdu comment to Roman Urdu
def to_roman_urdu(comment):
    # Ensure the comment is a string before splitting
    comment_str = str(comment)
    words = comment_str.split()
    roman_words = [roman_dict.get(w, w) for w in words]
    return ' '.join(roman_words)

# Step 4: Minor variations for augmentation
def augment_roman(text):
    words = text.split()

    # Randomly swap two words (10% chance)
    if len(words) > 2 and random.random() < 0.1:
        i, j = random.sample(range(len(words)), 2)
        words[i], words[j] = words[j], words[i]

    # Randomly replace 'bohat' with 'bahut'
    words = ['bahut' if w=='bohat' and random.random()<0.5 else w for w in words]

    # Randomly replace 'achhi' with 'acchi'
    words = ['acchi' if w=='achhi' and random.random()<0.5 else w for w in words]

    return ' '.join(words)

# Step 5: Create Roman Urdu variants
roman_comments = []
sentiments = []

for idx, row in df.iterrows():
    roman = to_roman_urdu(row['comment'])
    augmented = augment_roman(roman)

    roman_comments.append(augmented)
    sentiments.append(row['sentiment'])

# Step 6: Create DataFrame for augmented data
df_roman = pd.DataFrame({
    'comment': roman_comments,
    'sentiment': sentiments
})

# Step 7: Combine with original dataset
df_augmented = pd.concat([df, df_roman], ignore_index=True)

# Step 8: Save final augmented CSV
df_augmented.to_csv('augmented_urdu_dataset.csv', index=False)
print("Augmented dataset saved as 'augmented_urdu_dataset.csv' with", len(df_augmented), "rows")
```
