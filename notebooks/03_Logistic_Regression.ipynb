```python
urdu_df = pd.read_csv("final_urdu_dataset.csv")

urdu_tfidf = joblib.load("urdu_tfidf.pkl")

print("Urdu dataset shape:", urdu_df.shape)
print("Urdu TF-IDF loaded successfully!")
```

```python
X_urdu = urdu_tfidf.transform(
    urdu_df["comment"].fillna("")
)

y_urdu = urdu_df["sentiment"]

print("Urdu TF-IDF Shape:", X_urdu.shape)
print("Urdu labels:")
print(y_urdu.value_counts())
```

```python
X_train_urdu, X_test_urdu, y_train_urdu, y_test_urdu = train_test_split(
    X_urdu,
    y_urdu,
    test_size=0.20,
    random_state=42,
    stratify=y_urdu
)

print("Urdu Training data:", X_train_urdu.shape)
print("Urdu Testing data:", X_test_urdu.shape)
```

```python
urdu_model = LogisticRegression(
    max_iter=1000,
    random_state=42
)

urdu_model.fit(
    X_train_urdu,
    y_train_urdu
)

print("Urdu Logistic Regression trained successfully!")
```

```python
y_pred_urdu = urdu_model.predict(X_test_urdu)

print("Urdu prediction completed!")
```

```python
urdu_accuracy = accuracy_score(y_test_urdu, y_pred_urdu)
urdu_precision = precision_score(
    y_test_urdu, y_pred_urdu, average="weighted"
)
urdu_recall = recall_score(
    y_test_urdu, y_pred_urdu, average="weighted"
)
urdu_f1 = f1_score(
    y_test_urdu, y_pred_urdu, average="weighted"
)

print("Urdu Accuracy :", urdu_accuracy)
print("Urdu Precision:", urdu_precision)
print("Urdu Recall   :", urdu_recall)
print("Urdu F1 Score :", urdu_f1)

print("\nClassification Report:")
print(classification_report(y_test_urdu, y_pred_urdu))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test_urdu, y_pred_urdu))
```

```python
joblib.dump(
    urdu_model,
    "urdu_logistic_regression.pkl"
)

print("Urdu model saved successfully!")
```
