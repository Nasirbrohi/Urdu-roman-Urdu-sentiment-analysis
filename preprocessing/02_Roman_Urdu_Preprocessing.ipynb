Lowercasing

```python
# If there are NaN values, .str.lower() handles them well
romanurdu['comment'] = romanurdu['comment'].str.lower()
romanurdu['sentiment'] = romanurdu['sentiment'].str.lower()

# Check results
print(romanurdu[['comment', 'sentiment']].head())
```

Remove HTML Tags


```python
def remove_html_tags(text):
    """Remove HTML tags from text"""
    if not isinstance(text, str):
        return ""

    pattern = re.compile('<.*?>')
    return pattern.sub(r'', text)

# Apply to dataframe
romanurdu['comment'] = romanurdu['comment'].apply(remove_html_tags)

# Check results
print(romanurdu[['comment']].head())
```

Rmove urls

```python
import re

def remove_urls(text):
    """Remove URLs from text"""
    if not isinstance(text, str):
        return ""

    pattern = re.compile(r'http\S+|www\S+|https\S+')
    return pattern.sub(r'', text)

# Apply to dataframe
romanurdu['comment'] = romanurdu['comment'].apply(remove_urls)

# Check results
print(romanurdu[['comment']].head())
```

Punctuations

```python
import re
import string

def remove_punctuation(text):
    """Remove English and common punctuation from Roman Urdu text"""
    if not isinstance(text, str):
        return ""

    # Remove all English punctuation
    pattern = re.compile(r'[!"#$%&\'()*+,-./:;<=>?@[\\\]^_`{|}~]')
    return pattern.sub(' ', text)

# Apply to dataframe
romanurdu['comment'] = romanurdu['comment'].apply(remove_punctuation)

# Check results
print(romanurdu[['comment']].head())
```

Chatwords treatment

```python
chat_words = {
    "u": "you",
    "r": "are",
    "ur": "your",
    "k": "kay",
    "ki": "ki",
    "ka": "ka",
    "pls": "please",
    "plz": "please",
    "ppl": "people",
    "msg": "message",
    "tmrw": "tomorrow",
    "gm": "good morning",
    "gn": "good night",
    "tc": "take care",
    "bt": "but",
    "bcoz": "because",
    "bc": "because",
    "fr": "for",
    "4": "for",
    "gr8": "great",
    "m": "main",
    "me": "main",
    "hn": "haan",
    "ha": "haan",
    "h": "haan",
    "ap": "aap",
    "thx": "thanks",
    "tnx": "thanks",
    "gud": "good",
    "nyc": "nice",
    "bro": "bhai",
    "sis": "behen",
    "lol": "laugh",
    "omg": "oh my god",
    "idk": "i dont know",
    "ikr": "i know right",
    "wtf": "what the f",
    "wth": "what the hell",
    "smh": "shaking my head",
    "afaik": "as far as i know",
    "imo": "in my opinion",
    "jk": "just kidding"
}

def chat_conversion(text):
    if not isinstance(text, str):
        return ""

    new_text = []
    for w in text.split():
        if w.lower() in chat_words:  # Changed to .lower()
            new_text.append(chat_words[w.lower()])
        else:
            new_text.append(w)
    return " ".join(new_text)

romanurdu['comment'] = romanurdu['comment'].apply(chat_conversion)
```

Handling Emojis

```python
import re

def remove_emoji(text):
    """Remove emojis from Roman Urdu text"""
    if not isinstance(text, str):
        return ""

    emoji_pattern = re.compile("["
                           u"\U0001F600-\U0001F64F"  # emoticons
                           u"\U0001F300-\U0001F5FF"  # symbols & pictographs
                           u"\U0001F680-\U0001F6FF"  # transport & map symbols
                           u"\U0001F1E0-\U0001F1FF"  # flags (iOS)
                           u"\U00002702-\U000027B0"
                           u"\U000024C2-\U0001F251"
                           "]+", flags=re.UNICODE)
    return emoji_pattern.sub(r'', text)

# Apply to dataframe
romanurdu['comment'] = romanurdu['comment'].apply(remove_emoji)

# Check results
print(romanurdu[['comment']].head())
```

Tokenization

```python
import nltk
from nltk.tokenize import word_tokenize
nltk.download('punkt')
nltk.download('punkt_tab')
romanurdu["tokens"] = romanurdu["comment"].apply(lambda x: word_tokenize(str(x)))
```
