Remove HTML **Tags**

```python
import re
def remove_html_tags(text):
      if not isinstance(text, str):
        return ""
      pattern = re.compile('<.*?>')
      return pattern.sub(r'', text)
```

```python
urdu['comment'] = urdu['comment'].apply(remove_html_tags)
```

Remove **URLs**

```python
def remove_urls(text):
    if not isinstance(text, str):
        return ""

    # Handles http, https, www, and common URL patterns
    pattern = re.compile(r'https?://\S+|www\.\S+')
    return pattern.sub(r'', text)

urdu['comment'] = urdu['comment'].apply(remove_urls)
```

**Punctuations**

```python
def remove_punctuation(text):
    if not isinstance(text, str):
        return ""

    # Remove everything except ۔ (Urdu full stop)
    pattern = re.compile(r'[!"#$%&\'()*+,-./:;<=>?@[\\\]^_`{|}~،؛؟]')
    return pattern.sub(' ', text)

urdu['comment'] = urdu['comment'].apply(remove_punctuation)
```

Chatwords **treatment**


```python
urdu_chatwords = {
    "plz": "براہ مہربانی",
    "pls": "براہ مہربانی",
    "tnx": "شکریہ",
    "thx": "شکریہ",
    "ws": "وعلیکم السلام",
    "as": "السلام علیکم",
    "asl": "السلام علیکم",
    "g": "جی",
    "ya": "یا",
    "uae": "متحدہ عرب امارات",
    "u": "آپ",
    "r": "ہیں",
    "bt": "لیکن",
    "bcoz": "کیوں کہ",
    "bc": "کیوں کہ",
    "fr": "کے لیے",
    "4": "کے لیے",
    "m": "میں",
    "me": "میں",
    "hn": "ہاں",
    "ha": "ہاں",
    "nahi": "نہیں",
    "nai": "نہیں",
    "ni": "نہیں",
    "q": "کیوں",
    "kn": "کچھ نہیں",
    "gd": "اچھا",
    "nyc": "اچھا",
    "awsm": "زبردست",
    "hru": "آپ کیسے ہیں",
    "wyd": "کیا کر رہے ہو",
    "idk": "مجھے نہیں معلوم",
    "omg": "اوہ میرے خدا",
    "lol": "ہنسی",
    "jk": "مزاق کر رہا ہوں",
    "kb": "کب",
    "abhi": "ابھی",
    "agr": "اگر",
    "mgr": "مگر",
    "lekin": "لیکن",
    "phr": "پھر",
    "kl": "کل",
    "aj": "آج",
    "mj": "مجھے",
    "tm": "تم",
    "ap": "آپ",
    "bh": "بھی",
    "bs": "بس"
}
```

```python
def chat_conversion(text):
    if not isinstance(text, str):
        return ""

    new_text = []
    for w in text.split():
        if w.upper() in urdu_chatwords:
            new_text.append(urdu_chatwords[w.upper()])
        else:
            new_text.append(w)
    return " ".join(new_text)

urdu['comment'] = urdu['comment'].apply(chat_conversion)
```

```python
urdu['comment'] = urdu['comment'].apply(chat_conversion)
```

Handling Emojis

```python
import re

def remove_emoji(text):
    if not isinstance(text, str):
        return ""

    emoji_pattern = re.compile("["
                           u"\U0001F600-\U0001F64F"  # emoticons
                           u"\U0001F300-\U0001F5FF"  # symbols & pictographs
                           u"\U0001F680-\U0001F6FF"  # transport & map symbols
                           u"\U0001F1E0-\U0001F1FF"  # flags (iOS)
                           u"\U00002702-\U000027B0"
                           u"\U000024C2-\U0001F251"
                           "]+", flags=re.UNICODE)
    return emoji_pattern.sub(r'', text)

urdu['comment'] = urdu['comment'].apply(remove_emoji)
```
